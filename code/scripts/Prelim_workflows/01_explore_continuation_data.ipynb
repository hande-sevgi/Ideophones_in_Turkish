```python
## Setting the files

import pandas as pd

prelim_data = pd.read_csv(
    "PreliminaryStudyContinuationTask_long_utf8.csv"
)

print("The data were loaded successfully.")
```

```python
## Data inspection

prelim_data.shape

prelim_data.head()
prelim_data.columns.tolist()


print(prelim_data['participant_id'].nunique())

print(prelim_data['participant_id'].value_counts().sort_index())

print(prelim_data['condition_code'].value_counts())
```

```python
## General analysis/sanity check

summary_continuation = (
    prelim_data.groupby(['polarity', 'continuation_type'])
    .size()
    )

summary_continuation
```

```python
## Analysis

prelim_analysis = prelim_data[
    ['participant_id',
    'condition_code',
    'response_text']
]

prelim_analysis.isna().sum()

prelim_analysis['response_text'].str.strip().eq(" ").sum()

neg_manner = prelim_analysis[
    prelim_analysis["condition_code"] == "neg_manner"][
        ['participant_id', 'response_text']
    ]

print(neg_manner)

neg_temporal = prelim_analysis[
    prelim_analysis['condition_code'] == "neg_temporal"][
      ['participant_id', 'response_text']
    ]

print(neg_temporal)
```
